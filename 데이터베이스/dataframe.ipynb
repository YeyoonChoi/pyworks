```python
import pandas as pd

# 1차원 리스트 데이터 생성 - Series 클래스
data = ['식빵', '딸기잼', '와인']
s1 = pd.Series(data, index=[1, 2, 3])
print(s1)
print(s1.index)
print(s1.values) # ['식빵', '딸기잼', '와인']
print(s1.shape) #(3,)
```

데이터프레임(DataFrame) - 행과 열로 이루어진 2차원 데이터 구조이다

```python
import pandas as pd

# 데이터프레임 생성
df = pd.DataFrame({
    "name": ['키보드', '마우스', 'USB 메모리'],
    "price": [20000, 32000, 11000]
})
print(df)
print(df.columns)
print(df.shape)

# 검색
print(df['name']) # 검색대괄호 []
print(df[['name', 'price']]) # 리스트와 검색대괄호 []. 2개 이상일 때 리스트를 한 번 더 감싸준다

# 칼럼 이름을 한글로 작성
# df.rename(columns={"name":"품명", "price":"가격"}, inplace=True) #inplace=True 즉시저장
df = df.rename(columns={"name":"품명", "price":"가격"})
print(df)
```

```python
# 칼럼 추가
df['사양(spec)'] = ["유선", "무선", "128GB"]

print(df)
```
