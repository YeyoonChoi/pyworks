### 시리즈(Series) : 1차원 배열의 자료 구조(파이썬의 리스트와 유사)
- 시리즈는 칼럼이 1개 존재

```python
import pandas as pd

data = ['수학', '영어', '과학']
s1 = pd.Series(data)
print(s1)
print(s1.values) #['수학', '영어', '과학']
print(s1.index)
print(s1.shape) #(3,) 튜플
```

### 데이터프레임(DataFrame) - 행(row)와 열(column)의 2차원 표 형태의 자료구조
- 칼럼이 2개 이상 존재

```python
import pandas as pd

df = pd.DataFrame({
    "품명" : ['키보드', '마우스', 'USB 메모리'],
    "가격" : [20000, 32000, 11000,]
})
print(df)
print(df.columns) #칼럼
print(df.index) #인덱스
print(df.shape)

# 칼럼 추가
df["사양"] = ["기계식", "광학", "128GB"]
print(df)
```

#### CSV 파일로 만들기 - to_csv(파일경로)
#### CSV 파일 읽기 - read_csv(파일경로)

```python
# 데이터프레임을 csv 파일로 생성
df.to_csv("datas/computer.csv", index=False)

# csv 파일 읽기
com = pd.read_csv("datas/computer.csv")

# 인덱스 변경
com.index = [1, 2, 3]
print(com)

# csv 파일로 수정
com.to_csv("datas/computer.csv", index=False)
```

```python
# 열 검색
com["품명"]
com[["품명", "가격"]]

# 행 검색 - Loc(행번호)
# com[1] - 오류
com.loc[1] # 첫째 행 선택
com.loc[1:2] # 첫째행과 둘째행

# 특정 값 선택 - Loc(행번호, 열 이름)
print(com.loc[1, "품명"]) #키보드
print(com.loc[2, "가격"]) #32000
print(com)
```

```python
# 요소값 수정
com.loc[3, "사양"] = "64GB"

# 행 추가
com.loc[4] = ["모니터", 83000, "24인치"]

# 행 삭제 - dro(행이름, axis=0)
com.drop(4, axis=0, inplace=True)

print(com)
```

```python
# 열 추가
com["할인가"] = com["가격"] * 0.9

# 열 삭제 - drop(열이름, axis=1)
com.drop("할인가", axis=1, inplace=True)
print(com)
```
